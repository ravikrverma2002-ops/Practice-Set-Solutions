```python
# Marks in five subjects. calculating the total, percentage and grade.


marks1 = float(input("Enter marks of sub1: "))
marks2 = float(input("Enter marks of sub2: "))
marks3= float(input("Enter marks of sub3: "))
marks4= float(input("Enter marks of sub4: "))
marks5 = float(input("Enter marks of sub5: "))

total= marks1 + marks2 + marks3 + marks4 + marks5

percentage = round(total / 500 * 100,2)

if percentage >= 80:
    grade = "A"
elif percentage >= 65:
    grade = "B"
elif percentage >= 50:
    grade = "C"
elif percentage >= 35:
    grade = "D"
else:
    grade = "F"

print("total:",total)
print("percentage:",percentage)
print("grade:",grade)
```

```python
# Calculate an electricity bill

units = int(input("Enter electricity units: "))

if units <= 100:
    bill = units * 5 

elif units <= 200:
    bill = (100 * 5) + ((units - 100) * 7)

else:
    bill = (100 * 5) + (100 * 7) + ((units - 200) * 10)

print("Electricity bill: ", bill)
```

```python
# Whether a number is armstrong number

num = int(input("Enter a number: "))

original = num
digits = len(str(num))
total = 0 

while num > 0:
    digit = num % 10
    total = total + digit ** digits
    num = num // 10

if total == original:
    print("Armstrong number")
else:
    print("Not an Armstrong number")
```

```python
# Count the no. of vowels, consonant, digits and special characters in a string.

t = input("Enter a string: ")

vowels = 0 
consonant = 0 
digits = 0 
special = 0 

for ch in t:
    if ch in "aeiou" or ch in "AEIOU":
        vowels += 1
    elif ch.isalpha():
        consonant += 1
    elif ch.isdigit():
        digits += 1
    else:
        special += 1

print("vowels:",vowels)
print("consonant:",consonant)
print("digits:",digits)
print("special characters:", special)
```

```python
# Whether a str is palindrome using slicing.

a = input("Enter a string: ")

if a == a[::-1]:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

```python
# Find the frequency of each character.

x = input("Enter a string: ")

frequency = {}

for ch in x:
    if ch in frequency:
       frequency[ch] += 1
    else:
        frequency[ch] = 1 
print(frequency)
```

```python
# All prime no. between 1 & N 

n = int(input("Enter N:"))

for num in range(2, n + 1):
    is_prime = True

    for i in range(2, num):
        if num % i == 0:
           is_prime = False
           break

    if is_prime:
       print(num)
```

```python
#Check whether a number is a Perfect number 

num = int(input("Enter a number: "))

total = 0 

for i in range(1, num): 
    if num % i == 0: 
        total += i 

if total == num: 
    print("Perfect number")
else: 
    print("Not a perfect number")
```

```python
# Floyd's Triangle up to N rows 

n = int(input("Enter a number: "))

num = 1 

for i in range(1, n + 1): 
    for j in range(i): 
        print(num, end=" ")
        num += 1

    print()
```

```python
# Largest, Smallest and average in a list

numbers = list(map(int, input("Enter number: ").split()))

largest = max(numbers)
smallest = min(numbers)
average = sum(numbers) / len(numbers)

print("Largest: ", largest)
print("Smallest:", smallest)
print("Average:", average)
```

```python
# Remove duplicate elements while maintaining original order

nums = list(map(int, input("Enter nums: ").split()))

result = []

for num in nums: 
    if num not in result:
        result.append(num)

print("After removing duplicates:", result)
```

```python
# Second largest and second smallest in a list 

nums = list(map(int, input("Enter nums: ").split()))

nums = list(set(nums))
nums.sort()

print("Second largest:", nums[-2])
print("Second smallest:", nums[1])
```

```python
# Separate even/odd numbers

nums = list(map(int, input("Enter nums: ").split()))

even = []
odd = []

for n in nums: 
    if n % 2 == 0: 
        even.append(n)
    else: 
        odd.append(n) 

print("Even numbers:", even)
print("Odd numbers: ", odd)
```

```python
# Rotate a list by K positions to the left 

nums = list(map(int, input("Enter nums: ").split()))
k = int(input("Enter K: "))

k = k % len(nums)

rotated = nums[k:] + nums[:k]

print("Rotated list: ", rotated)
```
