```python
# A list containing only prime numbers between 1 & 100 using list comprehension 

prime = [num for num in range(2, 101) if all (num % i != 0 for i in range(2, int(num ** 0.5) + 1))]

print(prime)
```

```python
# Functions to check whether a number is prime 

def is_prime(num): 
    if num < 2: 
        return False

    for i in range(2, num): 
        if num % i == 0: 
            return False 

    return True 

number = int(input("Enter a number: "))

if is_prime(number): 
    print("Prime number")
else: 
    print("Not a prime number")
```

```python
# Function to calculate GCD (Greatest Common Divisor) of two numbers 

def gcd(a, b): 
    while b != 0: 
        a, b = b, a % b 
    return a 

n1 = int(input("Enter first number: "))
n2 = int(input("Enter second number: "))

print("GCD: ", gcd(n1, n2))
```

```python
# Functions that returns the second largest element 

def second_largest(nums): 
    nums = list(set(nums))
    nums.sort()
    return nums[-2]

nums = list(map(int, input("Enter nums: ").split()))

print("Second largest:", second_largest(nums))
```

```python
# Function that returns the longest words 

def longest_word(sen): 
    words = sen.split()

    longest = words[0]

    for word in words: 
        if len(word) > len(longest): 
            longest = word 

    return longest

sen = input("Enter a sentence: ")

print("Longest word:", longest_word(sen))
```

```python
# Merge two list and remove duplicates 

def merge_lists(l1, l2):

    result = []

    for item in l1 + l2: 
        if item not in result: 
            result.append(item)

    return result 

l1 = [1, 4, 9, 10]
l2 = [10, 15, 17, 20]

print(merge_lists(l1, l2))
```

```python
# Find common elements between two lists without sets 

l1 = [1, 2, 3, 4]
l2 = [2, 1, 5, 9]
common = [] 

for item in l1: 
    if item in l2: 
        common.append(item)

print("Common elements: ", common)
```

```python
# Find the frequency of every element in a list 

nums = list(map(int, input("Enter the list: ").split())) 

frequency = {}

for n in nums: 
    if n in frequency: 
        frequency[n] += 1 
    else: 
        frequency[n] = 1

print(frequency)
```

```python
# Sort a list without using a .sort()

nums = list(map(int, input("Enter the list: ").split()))

for i in range(len(nums)): 

    for j in range(0, len(nums) - i - 1): 

        if nums[j] > nums[j + 1]:
            nums[j], nums[j + 1] = nums[j + 1], nums[j]

print("Sorted list: ", nums)
```

```python
# Safely divide two nums using exception handling 

try: 

    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))

    result = round(num1 / num2, 2) 

    print("Result: ", result)

except ZeroDivisionError: 
    print("Cannot divide by zero.")

except ValueError: 
    print("Please enter valid numbers.")
```

```python
# Menu-driven calculator with exception handling 

while True: 

    print("\n-----Calculator-----")
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "5": 
        print("Calculator closed.")
        break

    try: 
        num1 = float(input("Enter first number: "))
        num2 = float(input("enter second number: "))

        if choice == "1": 
            print("Result:", num1 + num2)

        elif choice == "2": 
            print("Result:", num1 - num2)

        elif choice == "3": 
            print("Result:", num1 * num2)

        elif choice == "4": 
            try:
                print("Result:", num1 / num2)
            
            except  ZeroDivisionError: 
                print("Cannot divide by zero.")
        else: 
            print("Invalid choice.")

    except ValueError: 
        print("Please enter valid numbers.")
```
