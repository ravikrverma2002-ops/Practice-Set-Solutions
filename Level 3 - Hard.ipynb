```python
# Factorial using recursion

def factorial(n): 
    if n == 0 or n == 1: 
        return 1

    return n * factorial(n - 1)

num = int(input("Enter a number: "))
print("Factorial: ", factorial(num))
```

```python
# Fibonacci using recursion 

def fibonacci(n): 
    if n <= 1: 
        return n 

    return fibonacci(n - 1) + fibonacci(n - 2)

n = int(input("Enter number of terms: "))

for i in range(n): 
    print(fibonacci(i), end=" ")
```

```python
# Sum of digits using recursion 

def sum_digits(n): 
    if n == 0: 
        return 0

    return (n % 10) + sum_digits(n // 10)

num = int(input("Enter a number: "))
print("Sum of digits:", sum_digits(num))
```

```python
# Reverse a string using recursion 

def reverse_string(text):
    if len(text) <= 1: 
        return text
    return reverse_string(text[1:]) + text[0]

text = input("Enter a string: ")
print("Reversed:", reverse_string(text))
```

```python
# Tower of Hanoi using recursion 

def tower_of_hanoi(n, source, auxiliary, destination): 

    if n == 1: 
        print("Move disk 1 from", source, "to", destination)
        return 

    tower_of_hanoi(n - 1, source, destination, auxiliary)
    print("Move disk", n, "from", source, "to", destination)

    tower_of_hanoi(n - 1, source, destination, auxiliary)

n = int(input("Enter number of disks: "))

tower_of_hanoi(n, "A", "B", "C")
```

```python
# Recursion palindrome check 

def is_palindrome(text): 
    if len(text) <= 1:
        return True

    if text[0] != text[-1]: 
        return False

    return is_palindrome(text[1:-1])

text = input("Enter a string: ")

if is_palindrome(text): 
    print("Palindrome")
else: 
    print("Not a palindrome")
```

```python
# List comrehension + helper function for prime numbers 

# Helper function 

def is_prime(num): 
    if num <= 2: 
        return False

    for i in range(2, int(num ** 0.5) + 1): 
        if num % i == 0: 
            return False

    return True

# List comprehension 

numbers = list(range(1, 101))

primes = [num for num in numbers if is_prime(num)]

print(primes)
```

```python
# Read numbers until "stop" + statistics 

numbers = []

while True:

    value = input("Enter a number or 'stop': ")

    if value.lower() == "stop":
        break

    try:
        number = float(value)
        numbers.append(number)

    except ValueError:
        print("Invalid input. Please enter a number.")

if len(numbers) == 0:
    print("No numbers were entered.")

else:
    numbers.sort()

    minimum = min(numbers)
    maximum = max(numbers)
    average = sum(numbers) / len(numbers)

    n = len(numbers)

    if n % 2 == 1:
        median = numbers[n // 2]

    else:
        middle1 = numbers[n // 2 - 1]
        middle2 = numbers[n // 2]

        median = (middle1 + middle2) / 2

    print("Minimum:", minimum)
    print("Maximum:", maximum)
    print("Average:", average)
    print("Median:", median)
```

```python
# Flatten a nested list using recursion 

def flatten_list(nested):

    result = []

    for item in nested:

        if isinstance(item, list):
            result.extend(flatten_list(item))

        else:
            result.append(item)

    return result


nested = [1, [2, 3], [4, [5, 6]]]

print(flatten_list(nested))
```

```python
# Student Record Management System 

students = []


def add_student():
    try:
        roll = int(input("Enter roll number: "))
        name = input("Enter name: ")
        marks = float(input("Enter marks: "))

        student = {
            "roll": roll,
            "name": name,
            "marks": marks
        }

        students.append(student)

        print("Student added successfully.")

    except ValueError:
        print("Invalid input.")


def display_students():

    if not students:
        print("No student records found.")
        return

    for student in students:
        print(
            "Roll:", student["roll"],
            "| Name:", student["name"],
            "| Marks:", student["marks"]
        )


def search_student():

    try:
        roll = int(input("Enter roll number to search: "))

        result = [student for student in students
                  if student["roll"] == roll]

        if result:
            print(result[0])

        else:
            print("Student not found.")

    except ValueError:
        print("Invalid roll number.")


def save_students():

    try:
        with open("students.txt", "w") as file:

            for student in students:
                file.write(
                    f'{student["roll"]},{student["name"]},{student["marks"]}\n'
                )

        print("Records saved.")

    except Exception as e:
        print("Error:", e)


def load_students():

    try:
        with open("students.txt", "r") as file:

            for line in file:
                roll, name, marks = line.strip().split(",")

                students.append({
                    "roll": int(roll),
                    "name": name,
                    "marks": float(marks)
                })

        print("Records loaded.")

    except FileNotFoundError:
        print("No previous records found.")


while True:

    print("\n--- Student Record Management System ---")
    print("1. Add student")
    print("2. Display students")
    print("3. Search student")
    print("4. Save records")
    print("5. Load records")
    print("6. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        add_student()

    elif choice == "2":
        display_students()

    elif choice == "3":
        search_student()

    elif choice == "4":
        save_students()

    elif choice == "5":
        load_students()

    elif choice == "6":
        print("Program ended.")
        break

    else:
        print("Invalid choice.")
```
